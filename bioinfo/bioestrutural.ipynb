## Calculando a distância Euclidiana entre dois átomos

Agora que já realizamos o download de um arquivo de estrutura de proteína, vamos realizar algumas análises básicas nesse arquivo usando Biopython. No exemplo a seguir, vamos “navegar” pela estrutura da proteína, passando pelos modelos disponíveis, pelas cadeias, resíduos e, por fim, por cada átomo. Vamos ainda calcular a distância Euclidiana entre dois átomos. Para ilustrar vamos usar os carbonos-alfa dos resíduos 475 e 476

```python
# Distância Euclidiana entre átomos com Biopython
from Bio.PDB import *

parser = MMCIFParser()
estrutura = parser.get_structure('4mdp', 'md/4mdp.cif')

print(estrutura)

# estrutura -> modelos -> cadeias -> resíduos -> átomos

for modelo in estrutura:
  print(modelo)
  for cadeia in modelo:
    print(cadeia)
    for residuo in cadeia:
      nome = residuo.get_resname()
      if nome != 'HOH':
        print(nome, residuo.id[1])

        # Detalhes dos átomos

        #for atomo in residuo:

          #print(atomo.id) # id

          #print(atomo.coord) # coordenadas x, y, z



# distancia euclidiana entre lys 475 e leu 468 - ca

R1 = estrutura[0]['A'][475]['CA']
R2 = estrutura[0]['A'][476]['CA']

distancia = R1 - R2

print(distancia, 'angstrons')
```

Observe que as linhas que exibem detalhes dos átomos foram comentadas. Isso foi feito apenas para facilitar a visualização do resultado final. Caso o símbolo de comentário (#) seja removido dessas linhas, informações dos átomos, como o código identificador do átomo e suas coordenadas, seriam exibidas abaixo de cada resíduo.

# Input/output
## Como eu crio um objeto de estrutura a partir de um arquivo PDB?

```python
# Leitura de um arquivo PDB

# Cria um objeto responsável por interpretar arquivos PDB
parser = PDBParser()

# Lê a estrutura da proteína a partir de um arquivo PDB
# Substitua o caminho abaixo pelo caminho do seu arquivo
estrutura = parser.get_structure(
    "Minha_Proteina",
    r"1FAT.pdb"
)

print("Estrutura carregada com sucesso!")
```

```python
#Como eu crio um objeto de estrutura a partir de um arquivo mmCIF?

# Cria um objeto responsável por interpretar arquivos mmCIF
parser = MMCIFParser()

# Lê a estrutura da proteína a partir de um arquivo mmCIF
estrutura = parser.get_structure(
    "Minha_Proteina",
    r"estrutura_cofactors.cif"
)

print("Estrutura carregada com sucesso!")
```

```python
#Toda estrutura no Bio.PDB segue a arquitetura
#Structure
#    │
 #   ├── Model
  #          │
   #         ├── Chain
    #                │
     #               ├── Residue
      #                      │
       #                     ├── Atom
```

## Vamos explorar a estrutura

```python
for modelo in estrutura:
    print(modelo)
for modelo in estrutura:
    for cadeia in modelo:
        print(cadeia.id)
for residuo in estrutura.get_residues():
    print(residuo)
for atomo in estrutura.get_atoms():
    print(atomo)
```

```python
#Calculando distâncias


#Selecionando dois átomos
ca1 = estrutura[0]["A"][100]["CA"]

ca2 = estrutura[0]["A"][150]["CA"]

#Calculando

distancia = ca1 - ca2

print(distancia)
#Resultado em Ångströms
```

```python
#Obtendo a sequência da proteína
from Bio.PDB.Polypeptide import PPBuilder

ppb = PPBuilder()

for pp in ppb.build_peptides(estrutura):

    print(pp.get_sequence())
```

```python
# Salvando um novo arquivo PDB
from Bio.PDB import PDBIO

io = PDBIO()

io.set_structure(estrutura)

io.save("saida.pdb")
```
