Biopython fornece o módulo Bio.PDB que permite ler, manipular, analisar e exportar de estruturas de macromoléculas diretamente do ``Protein Data Bank (PDB)``. Para isso, deve-se inicialmente importar o módulo ``Bio.PDB`` com o código:

## A Arquitetura SMCRA (Hierarquia do Biopython)

Toda estrutura tridimensional no ``Bio.PDB`` é organizada internamente segundo o modelo SMCRA. Entender essa hierarquia é a chave para navegar pelos arquivos:  $$\text{Structure} \longrightarrow \text{Model} \longrightarrow \text{Chain} \longrightarrow \text{Residue} \longrightarrow \text{Atom}$$
- Structure: A macromolécula completa carregada do arquivo PDB/mmCIF.
-   Model: Um modelo tridimensional específico (ex: conformações ou estados de RMN).
-   Chain: A cadeia polipeptídica ou nucleica (ex: Cadeia 'A', Cadeia 'B').
-   Residue: O aminoácido, nucleotídeo ou molécula ligante (ex: Lys 475, HOH).
-   Atom: O átomo individual com suas coordenadas 3D (ex: CA, N, O).  

## Preparação do Ambiente e Download de Arquivos

Para começar, importamos as bibliotecas e fazemos o download direto da estrutura do PDB pelo código de identificação (ex: ``4MDP`` ou ``1FAT``).

```python
from Bio.PDB import PDBList, PDBParser, MMCIFParser, PDBIO
from Bio.PDB.Polypeptide import PPBuilder
```

```python
pdbl = PDBList()
pdbl.retrieve_pdb_file('4MDP', pdir='.', file_format='mmCif')
```

## Leitura e Parsing de Arquivos (PDB vs mmCIF)

Para carregar os dados na memória, utilize o parser adequado para a extensão do seu arquivo (``.pdb`` ou ``.cif``):

```python
parser_pdb = PDBParser(QUIET=True)

estrutura_pdb = parser_pdb.get_structure("Minha_Proteina", "1FAT.pdb")
print("Estrutura PDB carregada com sucesso!Eba!")
```

## Carregando arquivos formato mmCIF (.cif)

```python
parser_cif = MMCIFParser(QUIET=True)

estrutura = parser_cif.get_structure("4mdp", "4mdp.cif")
print("Estrutura mmCIF carregada com sucesso!Eba!")
```

Observe que as linhas que exibem detalhes dos átomos foram comentadas. Isso foi feito apenas para facilitar a visualização do resultado final. Caso o símbolo de comentário (#) seja removido dessas linhas, informações dos átomos, como o código identificador do átomo e suas coordenadas, seriam exibidas abaixo de cada resíduo.

## Navegando e Selecionando Elementos

Como a estrutura segue a ``arquitetura SMCRA``, você pode acessar elementos usando a sintaxe de colchetes ``[ ]`` do Python, como usado em dicionários e listas:

```python
# Passo a passo:
modelo  = estrutura[0] # Primeiro modelo (índice 0)
cadeia  = modelo["A"] # Cadeia A
residuo = cadeia[100] # Resíduo na posição número 100
ca      = residuo["CA"] # Átomo de Carbono-Alfa (CA)
```

```python
ca_direto = estrutura[0]["A"][100]["CA"]
print("Coordenadas do CA:", ca_direto.get_coord())
```

**Iterando sobre a Hierarquia**

```python
# 1. Metodo 1 - Percorrendo elemento por elemento
for modelo in estrutura:
    for cadeia in modelo:
        for residuo in cadeia:
            if residuo.get_resname() != 'HOH':  # Ignora moléculas de água
                print(f"Resíduo: {residuo.get_resname()} | Posição: {residuo.id[1]}")
```

```python
# 2. Métodos 2 - Direto comaAtalhos do Biopython
for residuo in estrutura.get_residues():
    print(residuo.get_resname(), residuo.id)

for atomo in estrutura.get_atoms():
    print(atomo.get_name(), atomo.get_coord())
```

## Extração de Propriedades dos Átomos

Cada objeto ``Atom`` contém informações físico-químicas e geométricas importantes para a análise da estrutura molecular:

| **Propriedade** | **Método Biopython** | **Descrição** |
|---|---|---|
| **Nome do átomo** | `atomo.get_name()` ou `atomo.id` | Identificador do átomo (ex.: `CA`, `N`, `CB`). |
| **Coordenadas** | `atomo.get_coord()` | Matriz NumPy contendo as coordenadas tridimensionais $(x, y, z)$ do átomo. |
| **B-Factor** | `atomo.get_bfactor()` | Fator de temperatura, relacionado à flexibilidade ou desordem estrutural do átomo. |
| **Ocupância** | `atomo.get_occupancy()` | Fração de ocupação da posição do átomo na estrutura cristalográfica. |

```python
for atomo in estrutura.get_atoms():
    print(f"Átomo: {atomo.get_name()}")
    print(f"Coordenadas (x,y,z): {atomo.get_coord()}")
    print(f"B-Factor: {atomo.get_bfactor()}")
    print(f"Ocupância: {atomo.get_occupancy()}")
    print("-" * 20)
```

## Calculando a distância Euclidiana entre dois átomos

Agora que já realizamos o download de um arquivo de estrutura de proteína, vamos realizar algumas análises básicas nesse arquivo usando Biopython.
No exemplo a seguir, vamos “navegar” pela estrutura da proteína, passando pelos modelos disponíveis, pelas cadeias, resíduos e, por fim, por cada átomo. Vamos ainda calcular a distância Euclidiana entre dois átomos. Para ilustrar vamos usar os carbonos-alfa dos resíduos 475 e 476. 

O Biopython simplifica o cálculo da Distância Euclidiana permitindo usar diretamente o operador de subtração (-) entre dois objetos ``Atom``:  Fórmula da Distância Euclidiana:$$d(A, B) = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$

```python
R1 = estrutura[0]['A'][475]['CA']
R2 = estrutura[0]['A'][476]['CA']

distancia = R1 - R2

print(f"Distância entre Lys 475 e Leu 476: {distancia:} Ångströms")
```

## Extração da Sequência de Aminoácidos

Para extrair a sequência primária contida na estrutura 3D, usamos a classe ``PPBuilder`` (Polypeptide Builder)

```python
from Bio.PDB.Polypeptide import PPBuilder

ppb = PPBuilder()
for pp in ppb.build_peptides(estrutura):
    sequencia = pp.get_sequence()
    print(f"Sequência peptídica: {sequencia}")
```

## Exportando e Manipulando Arquivos PDB

Você pode salvar modificações ou extrair apenas partes específicas de uma estrutura para um novo arquivo ``.pdb`` usando o ``PDBIO``:

```python
from Bio.PDB import PDBIO

io = PDBIO()
io.set_structure(estrutura)
io.save("saida_completa.pdb")
```

### Isolar e salvar apenas uma cadeia

```python
# Isola a Cadeia A
cadeia_a = estrutura[0]["A"]

# Salva em um novo arquivo PDB
io_cadeia = PDBIO()
io_cadeia.set_structure(cadeia_a)
io_cadeia.save("apenas_cadeia_A.pdb")

print("Cadeia A exportada com sucesso!Eba!")
```
